```python
#🔍Stop Words

#gey the reserved stop words
from nltk.corpus import stopwords
reserved_stop_words = set(stopwords.words('english')) # set >> to get the unique vals
print(reserved_stop_words,end=" ")
```

```python
all_stop_words=list(reserved_stop_words.union(extra_stop_words))#Combine the reserved_stop_words set with another the extra_stop_words
print(all_stop_words,end=" ")
```

```python
#🔍Convert to list from string (Tokenization)

#Tokenization :the process of breaking down text into individual words

import nltk

# Download required NLTK data
nltk.download('punkt')
nltk.download('punkt_tab')

# Now your tokenization will work
txt = "Hello, World! How are you?"
txt = nltk.word_tokenize(txt)

print(txt)
```

```python
#Stemming
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()

txt = ['Running', 'runners', 'ran', 'runs']
txt = [stemmer.stem(word) for word in txt]
print(txt)
```

```python
#🔍Text preprocessing

def txt_preprocessing(txt):
    txt = txt.lower()#lowercase
    txt = re.sub(r"<.*?>", " ", txt) # Remove HTML tags
    txt = re.sub(r"[^a-zA-Z]", " ", txt)# Remove special characters and digits
    txt = nltk.word_tokenize(txt)# Tokenize text
    txt = [word for word in txt if word not in all_stop_words] # Remove stopwords  
    txt = [word for word in txt if len(word) >= 3] # Remove words less than three letters
    txt = [stemmer.stem(word) for word in txt]  # Stem words
    
   
    return " ".join(txt) #return to string
```

```python
#🔍CountVectorizer
#🔴 select idx (index of paper) as you want.

idx=4
from sklearn.feature_extraction.text import CountVectorizer
cnt_vct = CountVectorizer(max_features=3000, ngram_range=(2, 4))
word_cnt_vct = cnt_vct.fit_transform(select_paper_to_process(idx))
```

```python
#CountVectorizer converts a collection of text documents into a matrix of token counts, capturing the frequency of words or n-grams for text analysis.
#max_features=3000 limits the feature set to the top 8,000 most frequent words or n-grams, reducing dimensionality and highlighting key terms.
#ngram_range=(2, 4) extracts 2-grams to 4-grams from the text, capturing sequences of words to incorporate contextual information.
#fit_transform(papers) learns the vocabulary from the text data and converts it into a document-term matrix with feature counts.

#🔍TF-IDF Transformation:

from sklearn.feature_extraction.text import TfidfTransformer
tfidf = TfidfTransformer(smooth_idf=True, use_idf=True)
tfidf.fit(word_cnt_vct)
```

```python
#TfidfTransformer Converts raw term counts into TF-IDF scores to emphasize term importance by balancing individual document and corpus-wide frequencies.
#smooth_idf=True applies smoothing to the IDF calculation to avoid division by zero and ensure non-zero IDF scores for all terms.
#use_idf=True indicates that the IDF component should be included in the TF-IDF calculation.

#Keyword Extraction

#🔍Sorting Sparse Matrix Columns by Value and Index

def sort_(matrix):
    tuples = list(zip(matrix.col, matrix.data))
    sorted_tuples = sorted(tuples, key=lambda x: (-x[1], x[0]))
    return sorted_tuples
```

```python
#🔍Extracting Top N Features with Scores from Sorted Items
def top_N(feature_names, sorted_items, topn=10):
    top_items = sorted_items[:topn]
    results = {}
    for idx, score in top_items:
        feature_name = feature_names[idx]
        results[feature_name] = round(score, 3)
    return results
```

```python
#Parameters:

#   feature_names: List of feature names corresponding to the indices in sorted_items.
#   sorted_items: List of tuples where each tuple contains an index and a score, sorted by score.
#   topn: The number of top features to extract (default is 10).

#Returns:

#   A dictionary where keys are feature names and values are their corresponding scores

#🔍Extracting Top N Keywords from a Document Based on TF-IDF Scores
def get_keywords(idx, text_list, cnt_vct, tfidf, topn=10):
    tf_idf_vector = tfidf.transform(cnt_vct.transform(text_list))
    sorted_items = sort_(tf_idf_vector.tocoo())
    feature_names = cnt_vct.get_feature_names_out()
    keywords = top_N(feature_names, sorted_items, topn)
    return keywords
```

```python
#Parameters:

#   idx: Index of the document in the 'docs' list.
#   papers: List of documents (texts).
#   cnt_vct: CountVectorizer instance used for transforming the documents.
#   tfidf: TF-IDF transformer instance used for generating TF-IDF vectors.
#   topn: The number of top keywords to extract (default is 10).

#🔍Display Document Title, Abstract, and Top Keywords.

def print_results(idx, keywords, df):
    print("\n=====Title=====")
    print(df['title'][idx])
    print("\n=====Abstract=====")
    print(df['abstract'][idx])
    print("\n===Keywords===")
    for k in keywords:
        print(k, keywords[k])
```
